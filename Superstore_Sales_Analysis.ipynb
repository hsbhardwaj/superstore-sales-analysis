Further product-level analysis is required to identify the specific

### Product-Level Deep Dive

The Cubify CubeX 3D Printer was identified as the main contributor
to Cindy Stewart's loss.

Further analysis of all transactions for this product shows that
the two orders receiving a 70% discount generated a combined loss
of $9,239.97, with an average loss of $4,619.98 per order.

In comparison, the order receiving a 20% discount generated a
profit of $359.99.

This indicates a strong association between the 70% discount level
and negative profitability for this product. Excessive discounting
should therefore be reviewed for high-value products such as this
3D printer.

## Business Insights — Correlation Analysis

1. Sales and Profit have a moderate positive correlation of 0.48,

   profit.

   suggesting that higher discounts are associated with lower profit.

   associated with profitability in this dataset.

4. Sales and Quantity have a weak positive correlation of 0.20,
   indicating that higher quantities sold do not always result in
   proportionally higher sales.

5. Shipping Days and Profit have almost no correlation (-0.005),
   suggesting that shipping duration has little linear relationship
   with profit in this dataset.

Note: Correlation indicates association and does not prove causation.

```python
Q1 = df_clean['Profit'].quantile(0.25)
Q3 = df_clean['Profit'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

profit_outliers = df_clean[
    (df_clean['Profit'] < lower_bound) |
    (df_clean['Profit'] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Profit Outliers:", len(profit_outliers))
```

## Business Insights — Outlier Analysis

1. The IQR method identified 1,881 profit outliers. These
   transactions were retained because they represent legitimate
   business transactions rather than confirmed data errors.

2. Extreme negative-profit transactions are frequently associated
   with high discount levels, particularly 40%–80% discounts.

3. Machines and Binders appear frequently among the largest
   loss-making transactions.

4. Extreme positive-profit transactions are commonly associated
   with high-value products such as Copiers and Binders, often
   with 0% discount.

5. The outlier analysis reinforces the earlier finding that
   discounting is strongly associated with profitability.

6. Outliers should therefore be investigated as potential business
   opportunities or risks rather than automatically removed.

# Final Business Insights

## Key Findings

1. Sales growth does not always translate into proportional profit
   growth. Several high-sales products and customers generate
   relatively low or negative profitability.

2. Discounting is one of the most important profitability factors
   identified in the analysis. Discount and Profit Margin have a
   strong negative correlation of -0.86.

3. Tables are a major profitability concern, generating approximately
   $206.97K in sales but a loss of $17.73K.

4. Copiers are a strong profitability category, generating
   approximately $55.62K in profit with a 37.20% profit margin.

5. The Central region requires attention because it has the lowest
   regional profit margin at 7.92% and the highest average discount
   at 24.04%.

6. Several states with high average discounts, including Texas,
   Ohio, Pennsylvania and Illinois, show negative profitability.

7. Customer-level analysis shows that high-revenue customers are not
   necessarily the most profitable customers.

8. 2017 recorded the highest annual sales and order volume, while
   2016 recorded the strongest year-over-year sales and profit growth.

9. November generated the highest total sales across the four-year
   period, while December generated the highest total profit.

10. Extreme loss-making transactions are frequently associated with
    high discounts, while several highly profitable transactions
    involve high-value products sold with low or zero discounts.

## Business Recommendations

1. Review high discount levels, especially discounts of 40% or more,
   for products and regions with weak profitability.

2. Investigate loss-making sub-categories such as Tables, Bookcases
   and Supplies.

3. Review the discount strategy in the Central region.

4. Identify profitable product-customer combinations and prioritize
   them for future sales strategies.

5. Monitor high-value transactions carefully to balance revenue
   growth with profitability.

6. Use product-level and regional profitability metrics alongside
   sales when making business decisions.
